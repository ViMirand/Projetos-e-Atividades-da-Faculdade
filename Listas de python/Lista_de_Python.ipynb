```python
## questão 4
Peso = float(input("Digite seu peso: "))
Altura = float(input("Digite sua altura: "))
IMC = Peso/(Altura**2)

print("\n--- Tabela de Classificação do IMC ---")
print("IMC \t\t Classificação")
print("------------------------------------")
print("Menos de 18.5 \t Magreza")
print("18.5 a 24.9 \t Normal")
print("25.0 a 29.9 \t Sobrepeso")
print("30.0 a 34.9 \t Obesidade Grau I")
print("35.0 a 39.9 \t Obesidade Grau II")
print("40.0 ou mais \t Obesidade Grau III")
print("------------------------------------")

print(f"\nSeu IMC é: {IMC:.2f}")
if IMC < 18.5:
    print("Classificação: Magreza")
elif IMC < 25.0:
    print("Classificação: Normal")
elif IMC < 30.0:
    print("Classificação: Sobrepeso")
elif IMC < 35.0:
    print("Classificação: Obesidade Grau I")
elif IMC < 40.0:
    print("Classificação: Obesidade Grau II")
else:
    print("Classificação: Obesidade Grau III")
```

```python
## questão 5
Sal = float(input("Digite seu salário por hora: "))
HT = int(input("Digite suas horas trabalhadas: "))
VB = Sal * HT
IR = VB * 0.11
INSS = VB * 0.275
print("\n\nSalário Bruto = {:.2f}".format(VB))
print("Valor pago ao INSS = {:.2f}".format(INSS))
print("Valor pago ao IR = {:.2f}".format(IR))
print("Salário Líquido = {:.2f}".format(VB-IR-INSS))
```

```python
##questão 6
Sal = float(input("Digite seu salário por hora: "))
HT = int(input("Digite suas horas trabalhadas: "))
IR = int(input("Digite o valor do IR: "))
INSS = int(input("Digite o valor do INSS: "))
VB = Sal * HT
IR = VB * (IR/100)
INSS = VB * (INSS/100)
print("\n\nSalário Bruto = {:.2f}".format(VB))
print("Valor pago ao INSS = {:.2f}".format(INSS))
print("Valor pago ao IR = {:.2f}".format(IR))
print("Salário Líquido = {:.2f}".format(VB-IR-INSS))
```

```python
##questão 15

N = float(input("Digite um numero: "))
OP = input("\n\n\tMENU DE OPÇÕES\nA)Par pu ímpar?\nB)Inteiro ou Decimal?\nC)Positivo ou negativo?\n\nEscolha apenas um.\nResposta: ")

if OP == "A" or OP == "a":
  if N%2 == 0:
    print("O numero é par")
  else:
    print("O numero é impar")
elif OP == "B" or OP == "b":
  try:
    inteiro = int(N)
  except ValueError as err :
    print(err)
  if inteiro:
    print("O numero é inteiro")
  else:
    print("O numero é decimal")
elif OP == "C" or OP == "c":
  if N > 0:
    print("O numero é positivo")
  elif N < 0:
    print("O numero é negativo")
  else:
    print("O numero é neutro, N é nulo")
```

```python
## questão 21
ids = []
N = int(input("Digite a quantidade de pessoas: "))
try:
  if N>0:
    for i in range(N):
      idade = int(input("Digite a idade: "))
      if idade > 0:
        ids.append(idade)
      else:
        idade = int(input("Idade inválida! Digite novamente:"))
        ids.append(idade)
      media = sum(ids)/len(ids)
    if(media <= 25):
      print("Média:", media, "\nTurma Jovem")
    elif media <= 60:
      print("Média:", media, "\nTurma Adulta")
    else:
      print("Média:", media, "\nTurma Idosa")
except ValueError as err:
  print(err)
```
