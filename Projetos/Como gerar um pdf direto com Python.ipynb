```python
from fpdf import FPDF
from datetime import datetime

class PDF(FPDF):
    def header(self):
        # Título do Documento
        self.set_font('Helvetica', 'B', 15)
        self.cell(0, 10, 'Dicionário de Dados - NYC TLC FHVHV', 0, 1, 'C')
        self.set_font('Helvetica', 'I', 10)
        self.cell(0, 10, f'Gerado em: {datetime.now().strftime("%d/%m/%Y %H:%M")}', 0, 1, 'C')
        self.ln(5)

    def chapter_title(self, label):
        self.set_font('Helvetica', 'B', 12)
        self.set_fill_color(200, 220, 255)
        self.cell(0, 10, f" {label}", 0, 1, 'L', True)
        self.ln(4)

# 1. Configuração dos Dados (Tudo manual aqui)
# O JEITO CERTO:
metadados = [
    ["Atributo", "Valor"],
    ["Nome do Projeto", "NYC TLC - High Volume For-Hire Vehicle"],
    ["Formato Original", "Apache Parquet"],
    ["Versão", "1.0 - Baseline (pré-ETL)"],
    ["Data de Criação", "30/03/2026"],
    ["Status", "Em revisão"]
]

# Principais colunas do dataset para o demonstrativo
# Versão corrigida para o seu script:
colunas = [
    ["hvfhs_license_num", "STRING", "Não", "Licença TLC da plataforma (Uber, Lyft...)"],
    ["request_datetime", "TIMESTAMP", "Não", "Data/Hora da solicitação"],
    ["trip_miles", "DOUBLE", "Não", "Distância total em milhas"],
    ["base_passenger_fare", "DOUBLE", "Não", "Tarifa base paga ao motorista"],
    ["tips", "DOUBLE", "Não", "Gorjeta capturada eletronicamente"]
]
# 2. Geração do PDF
pdf = PDF()
pdf.add_page()
pdf.set_auto_page_break(auto=True, margin=15)

# Seção 1: Informações Gerais
pdf.chapter_title("1. Visão Geral do Dataset")
pdf.set_font("Helvetica", size=10)
for row in metadados:
    pdf.set_font("Helvetica", "B", 10)
    pdf.cell(45, 8, row[0], 1)
    pdf.set_font("Helvetica", "", 10)
    pdf.cell(100, 8, row[1], 1)
    pdf.ln()

pdf.ln(10)

# Seção 2: Tabela de Colunas
pdf.chapter_title("2. Dicionário de Colunas (Demonstrativo)")
pdf.set_font("Helvetica", "B", 9)
pdf.set_fill_color(180, 180, 180)

# Cabeçalho da Tabela
larguras = [45, 25, 20, 90]
titulos = ["Coluna", "Tipo", "Nulo", "Descrição"]
for i in range(len(titulos)):
    pdf.cell(larguras[i], 8, titulos[i], 1, 0, 'C', True)
pdf.ln()

# Linhas da Tabela
pdf.set_font("Helvetica", "", 9)
for row in colunas:
    pdf.cell(larguras[0], 7, row[0], 1)
    pdf.cell(larguras[1], 7, row[1], 1)
    pdf.cell(larguras[2], 7, row[2], 1)
    pdf.cell(larguras[3], 7, row[3], 1)
    pdf.ln()

# 3. Salvando o arquivo
nome_arquivo = "Dicionario_Manual_TLC.pdf"
pdf.output(nome_arquivo)
print(f"Sucesso! O arquivo '{nome_arquivo}' foi gerado.")
```
