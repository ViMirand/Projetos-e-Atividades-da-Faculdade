```python
import io
import zipfile
import pandas as pd
import requests
from tqdm import tqdm
from huggingface_hub import login, HfApi

# 🔑 TOKEN
HF_TOKEN = ""
login(HF_TOKEN)

# 📁 REPO
repo_id = "ViMirand/Sinasc-Brasil-2024a2024"

# 📄 ARQUIVO FINAL
arquivo_final = "SINASC_15_A_24_COMPILADO.csv"

# 🔄 CONTROLE HEADER
primeiro = True

# 📊 LOOP
for i in tqdm(range(15, 25), desc="Processando anos"):

    url = f'https://s3.sa-east-1.amazonaws.com/ckan.saude.gov.br/SINASC/csv/SINASC_20{i}_csv.zip'
    print(f"\n📥 Baixando {url}")

    try:
        resposta = requests.get(url, timeout=60)
        resposta.raise_for_status()
    except Exception as e:
        print(f"❌ Erro no download {i}: {e}")
        continue

    conteudo = io.BytesIO(resposta.content)

    with zipfile.ZipFile(conteudo) as zippado:
        nome_arquivo = zippado.namelist()[0]

        with zippado.open(nome_arquivo) as f:

            for chunk in pd.read_csv(
                f,
                sep=';',
                encoding='latin1',
                chunksize=200_000,  # 🔥 CONTROLE DE MEMÓRIA
                low_memory=False
            ):

                chunk.columns = chunk.columns.str.upper()

                # 💾 escreve direto no CSV (sem explodir RAM)
                chunk.to_csv(
                    arquivo_final,
                    mode='w' if primeiro else 'a',
                    header=primeiro,
                    index=False
                )

                primeiro = False

print("\n📦 Arquivo final gerado!")

# 🚀 UPLOAD
api = HfApi()

api.upload_file(
    path_or_fileobj=arquivo_final,
    path_in_repo=arquivo_final,
    repo_id=repo_id,
    repo_type="dataset"
)

print("✅ Upload concluído com sucesso!")
```
